```python
import pickle
import numpy as np
```

```python
names = ['activatedsludge',
         'Glacier',
         'GUT.ERP015450',
         'GUT.SRP056641',
         'Lake',
         'ORAL',
         'River',
         'Soil',]
```

```python
for name in names:
    filtered_name = 'filtered_datasets/full_dataset.filtered.Grilli.{}.dat'.format(name)
    unfiltered_name = 'unfiltered_datasets/full_dataset.Grilli.{}.dat'.format(name)

    with open(unfiltered_name,'rb') as f:
        data = pickle.load(f)
        f.close()
        M_pre,N_pre = np.shape(data)

    with open(filtered_name,'rb') as f:
        data = pickle.load(f)
        f.close()
        M_post,N_post= np.shape(data)

    print(name,'\n',
        M_pre,N_pre,'\n',
        M_post,N_post)
```
